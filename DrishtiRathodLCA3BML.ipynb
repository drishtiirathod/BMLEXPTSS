```python
# Step 1: Import libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score
import urllib.request

# Step 2: Load the dataset
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data'
columns = [
    'age', 'workclass', 'fnlwgt', 'education', 'education-num',
    'marital-status', 'occupation', 'relationship', 'race', 'sex',
    'capital-gain', 'capital-loss', 'hours-per-week', 'native-country', 'income'
]

df = pd.read_csv(url, header=None, names=columns, na_values=' ?', skipinitialspace=True)

# Step 3: Drop missing values
df.dropna(inplace=True)

# Step 4: Convert target to binary
df['income'] = df['income'].map({'<=50K': 0, '>50K': 1})

# Step 5: One-hot encode categorical variables
X = pd.get_dummies(df.drop('income', axis=1))
y = df['income'].values

# Step 6: Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 7: Apply K-means clustering
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_scaled)

# Step 8: Map cluster labels to actual income labels
mapped_labels = np.zeros_like(clusters)
for i in range(2):
    mask = (clusters == i)
    mapped_labels[mask] = np.bincount(y[mask]).argmax()

# Step 9: Calculate and print accuracy
accuracy = accuracy_score(y, mapped_labels)
print(f"K-means Clustering Accuracy on Adult Dataset: {accuracy * 100:.2f}%")
```
